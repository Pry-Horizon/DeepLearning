# 吴恩达深度学习环境测试

运行下面的单元格，验证所有库是否正常安装。

```python
# 导入所有核心库
import numpy as np
import matplotlib
import pandas as pd
import matplotlib.pyplot as plt
import scipy
import sklearn

print('✅ 基础库导入成功')
print(f'  numpy: {np.__version__}')
print(f'  pandas: {pd.__version__}')
print(f'  matplotlib: {matplotlib.__version__}')
print(f'  scipy: {scipy.__version__}')
print(f'  scikit-learn: {sklearn.__version__}')
```

```python
# 测试 numpy 运算
a = np.array([1, 2, 3, 4, 5])
b = np.array([5, 4, 3, 2, 1])

print(f'数组a: {a}')
print(f'数组b: {b}')
print(f'a + b = {a + b}')
print(f'a * b = {a * b}')
print(f'a的均值: {a.mean():.2f}')
print(f'a的标准差: {a.std():.2f}')
print('\n✅ numpy 运算正常')
```

```python
# 测试 matplotlib 画图
%matplotlib inline

x = np.linspace(0, 2 * np.pi, 100)
y_sin = np.sin(x)
y_cos = np.cos(x)

plt.figure(figsize=(10, 5))
plt.plot(x, y_sin, label='sin(x)', linewidth=2)
plt.plot(x, y_cos, label='cos(x)', linewidth=2)
plt.title('Sin & Cos Wave', fontsize=14)
plt.xlabel('x', fontsize=12)
plt.ylabel('y', fontsize=12)
plt.legend(fontsize=12)
plt.grid(True, alpha=0.3)
plt.show()

print('✅ matplotlib 画图正常')
```

```python
# 测试深度学习框架
try:
    import torch
    x = torch.tensor([1, 2, 3, 4, 5])
    print(f'✅ PyTorch {torch.__version__}')
    print(f'  tensor: {x}')
    print(f'  sum: {x.sum().item()}')
except ImportError:
    print('⚠️  PyTorch 未安装')

print()

try:
    import tensorflow as tf
    x = tf.constant([1, 2, 3, 4, 5])
    print(f'✅ TensorFlow {tf.__version__}')
    print(f'  tensor: {x.numpy()}')
    print(f'  sum: {tf.reduce_sum(x).numpy()}')
except ImportError:
    print('⚠️  TensorFlow 未安装')
```
