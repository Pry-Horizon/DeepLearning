```python
#导入库

import numpy as np        #矩阵计算
import h5py               #读取数据
import matplotlib.pyplot as plt     #画图
```

```python
#Sigmoid函数

def sigmoid(x):
    s = 1/(1+np.exp(-x))
    return s
```

```python
#测试

x = np.array([0,2])
print(sigmoid(x))
```

```python
#sigmoid 梯度

def sgmoid_derivative(x):
    s = sigmoid(x)
    ds = s*(1-s)
    return ds
```

```python
#重塑图片矩阵

def image2vector(image):
    v = image.reshape(image.shape[0]*image.shape[1]*image.shape[2],1)
    return v
```

```python
#标准化矩阵

def normalizeRows(x):
    x_norm = np.linalg.norm(x, axis=1, keepdims=True)
    x = x/x_norm
    return x
```

```python
#softmax函数

def softmax(x):

    x_exp=np.exp(x)

    x_sum=np.sum(
        x_exp,
        axis=1,
        keepdims=True
    )

    s=x_exp/x_sum

    return s
```
