```python
import numpy as np

a = np.array([1,2,3,4])
print(a)
```

```python
import time

a = np.random.rand(1000000)
b = np.random.rand(1000000)

#向量化计算，大幅提高计算效率
tic = time.time()
c = np.dot(a,b)
print(c)
toc = time.time()

print("Vectorized version: " + str(1000*(toc - tic)) + "ms")

#常规计算效率低
c = 0

tic = time.time()
for i in range(1000000):
    c += a[i]*b[i]
print(c)
toc = time.time()

print("For loop version: " + str(1000*(toc - tic)) + "ms")
```
